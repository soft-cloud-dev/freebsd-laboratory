# Welcome to the FreeBSD Laboratory
Executable documentation backed by a real FreeBSD runtime.

## System information
Discover the target FreeBSD runtime and capture its identity as evidence.

```python
import subprocess
for command in (["uname", "-a"], ["sysctl", "kern.ostype"], ["sysctl", "kern.osrelease"]):
    result = subprocess.run(command, capture_output=True, text=True, check=False)
    print(result.stdout.strip())
```
