## System information
Discover the target FreeBSD runtime and capture its identity as evidence.

```python
import subprocess
print(subprocess.run(["uname", "-a"], capture_output=True, text=True).stdout)
print(subprocess.run(["sysctl", "kern.ostype"], capture_output=True, text=True).stdout)
print(subprocess.run(["sysctl", "kern.osrelease"], capture_output=True, text=True).stdout)
```
