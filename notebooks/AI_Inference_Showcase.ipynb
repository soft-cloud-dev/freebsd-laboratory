## 3. In-Cell Q&A with `%ai` Line Magic

Ask questions about FreeBSD system architecture, networking, kernel tuning, or storage. The response is automatically rendered in rich GitHub-flavored Markdown:

```python
%ai Explain the architectural differences between VNET jails and bhyve virtual machines in FreeBSD.
```

## 4. Multi-Line Code Generation with `%%ai` Cell Magic

Use the `%%ai` cell magic to synthesize systems engineering code, configurations, or shell scripts.

### Supported Options:
- `--tokens <N>`: Maximum token budget (default: 512)
- `--temp <float>`: Temperature for sampling (default: 0.0 for deterministic output)
- `--system "<prompt>"`: Custom system instructions
- `--model <path>`: Specific model override

```python
%%ai --tokens 384
Write a Python script using ctypes to read sysctl kern.boottime on FreeBSD
and format the boot time as an ISO 8601 timestamp string.
```

## 5. Autonomous Troubleshooting with `%%agent` Magic

The `%%agent` magic launches an unprivileged autonomous agent that executes in an isolated bhyve VM or VNET jail runtime, diagnoses the requested task, and emits a structured report.

### Options:
- `--mode <bhyve|jail>`: Runtime isolation layer (default: `bhyve`)
- `--steps <N>`: Maximum execution step ceiling (default: 16)
- `--runtime <seconds>`: Overall execution deadline (default: 300)

```python
%%agent --mode bhyve --steps 8
Inspect the active network interfaces, IP addresses, and routing table inside the guest.
```
